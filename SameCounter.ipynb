```python
import re
import pandas as pd

df=pd.read_csv(r"C:/Users/yourname/Documents/DATASET.csv")

def normalize(text):
    if pd.isna(text):
        return ""
    text = str(text)
    # remove anything in brackets
    text = re.sub(r'\(.*?\)|\[.*?\]|\{.*?\}', '', text)
    # remove "limited" or "ltd" (case insensitive, as whole words)
    text = re.sub(r'\bltd\b', '', text, flags=re.IGNORECASE)
    text = re.sub(r'\blimited\b', '', text, flags=re.IGNORECASE)
    # remove all whitespace
    text = re.sub(r'\s+', '', text)
    return text.strip().lower()

# filter to relevant years. Change parameters here as needed. 
df_filtered = df[
    (df['Year'] >= 2015) & (df['Year'] <= 2024) & 
    (df['NO. OF GUARDS EMPLOYED'] > 50) & (df['NO. OF GUARDS EMPLOYED'] < 201)
    & (df['License Class'] == 'C' )
    ].copy()

# apply normalization
df_filtered['B_norm'] = df_filtered['Company Name'].apply(normalize)

# build a set of normalized names per year
year_sets = df_filtered.groupby('Year')['B_norm'].apply(set)

# compare consecutive years
years = sorted(year_sets.index)
results = {}
for y1, y2 in zip(years, years[1:]):
    common = year_sets[y1] & year_sets[y2]
    results[f"{y1}-{y2}"] = len(common)

results_df = pd.DataFrame(list(results.items()), columns=['Year_Pair', 'Matching_Count'])
print(results_df)
```
